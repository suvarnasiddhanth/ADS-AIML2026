Stack ADT Implementation

```python
class Stack:
  def __init__(self):
    self._items = []

  def is_empty(self):
    return len(self._items) == 0

  def push(self, item):
    self._items.append(item)

  def pop(self):
    if len(self._items) == 0:
      raise Exception("Stack is empty")
    else:
      return self._items.pop()

  def peek(self):
    if len(self._items) == 0:
      raise Exception("Stack is empty")
    else:
      return self._items[-1]

  def size(self):
    return len(self._items)

  def __str__(self):
    result = ""
    for item in range(len(self._items)-1, -1, -1):
      result += "|" + str(self._items[item]) + "|\n"
    return result
```

```python
stack = Stack()

print("Is stack empty?", stack.is_empty())

stack.push(1)
stack.push(2)
stack.push(3)
stack.push(4)

print("Stack after push: \n", stack)

stack.pop()

print("Stack after pop: \n", stack)

print("Top of stack:", stack.peek())

print("Stack size:", stack.size())
```

Stack ADT Implementation - Linked List

```python
class Node:
    """Represents a single node in the linked list."""
    def __init__(self, data):
        self.data = data
        self.next = None  # Points to the node underneath


class LinkedStack:
    """Implements a Stack ADT using a Linked List."""
    def __init__(self):
        self.top = None      # Points to the top element of the stack
        self._size = 0       # Tracks the number of elements

    def push(self, data):
        """Adds an element to the top of the stack. O(1) Time."""
        new_node = Node(data)
        new_node.next = self.top  # Link new node to the current top
        self.top = new_node       # Move top pointer to the new node
        self._size += 1

    def pop(self):
        """Removes and returns the top element. Raises IndexError if empty. O(1) Time."""
        if self.is_empty():
            raise IndexError("Stack Underflow: Cannot pop from an empty stack.")

        popped_node = self.top    # Keep track of old top
        self.top = self.top.next  # Move top pointer down
        self._size -= 1
        return popped_node.data

    def peek(self):
        """Returns the top element without removing it. O(1) Time."""
        if self.is_empty():
            raise IndexError("Stack is empty: Cannot peek.")
        return self.top.data

    def is_empty(self):
        """Returns True if the stack is empty, False otherwise. O(1) Time."""
        return self.top is None

    def size(self):
        """Returns the total number of items in the stack. O(1) Time."""
        return self._size

    def display(self):
        """Helper method to print the entire stack contents from top to bottom."""
        current = self.top
        elements = []
        while current:
            elements.append(str(current.data))
            current = current.next
        print(" -> ".join(elements) if elements else "Empty Stack")


# --- Driver Code to Demonstrate Functionality ---
if __name__ == "__main__":
    stack = LinkedStack()

    # Push elements
    stack.push(10)
    stack.push(20)
    stack.push(30)

    print("Current Stack (Top to Bottom):")
    stack.display()  # Outputs: 30 -> 20 -> 10

    print(f"Top element (peek): {stack.peek()}")  # Outputs: 30
    print(f"Stack size: {stack.size()}")          # Outputs: 3

    # Pop elements
    print(f"Popped: {stack.pop()}")               # Outputs: 30
    print("Stack after pop:")
    stack.display()  # Outputs: 20 -> 10
```
