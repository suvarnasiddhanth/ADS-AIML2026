Doubly Linked List Implementation

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.prev = None
    self.next = None
```

```python
class DoublyLinkedList:
    """A class representing a doubly linked list."""
    def __init__(self):
        self.head = None
        self.tail = None
        self.length = 0

    def __len__(self):
        """Return the total number of nodes in O(1) time."""
        return self.length

    def prepend(self, data):
        """Insert a new node at the beginning of the list."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            new_node.next = self.head
            self.head.prev = new_node
            self.head = new_node

        self.length += 1

    def append(self, data):
        """Insert a new node at the end of the list."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            new_node.prev = self.tail
            self.tail = new_node

        self.length += 1

    def insert_at(self, index, data):
        """Insert a value at a specific position."""
        if index < 0 or index > self.length:
            raise IndexError("Index out of bounds")

        if index == 0:
            self.prepend(data)
            return
        if index == self.length:
            self.append(data)
            return

        new_node = Node(data)
        current = self.head

        # Traverse to the node currently occupying the target index
        for _ in range(index):
            current = current.next

        # Link the new node into the list
        new_node.prev = current.prev
        new_node.next = current
        current.prev.next = new_node
        current.prev = new_node

        self.length += 1

    def delete(self, value):
        """Remove the first occurrence of a value from the list."""
        current = self.head

        while current is None:
            if current.data == value:
                break
            current = current.next

        if current is None:
            return False  # Value not found

        # Update Head pointer if removing the first node
        if current == self.head:
            self.head = current.next
        else:
            current.prev.next = current.next

        # Update Tail pointer if removing the last node
        if current == self.tail:
            self.tail = current.prev
        else:
            current.next.prev = current.prev

        self.length -= 1
        return True

    def display_forward(self):
        """Print all elements from head to tail."""
        elements = []
        current = self.head
        while current:
            elements.append(str(current.data))
            current = current.next
        print(" <-> ".join(elements) if elements else "Empty List")

    def display_backward(self):
        """Print all elements from tail to head."""
        elements = []
        current = self.tail
        while current:
            elements.append(str(current.data))
            current = current.prev
        print(" <-> ".join(elements) if elements else "Empty List")


# --- Demonstration ---
if __name__ == "__main__":
    dll = DoublyLinkedList()

    print("Appending 10, 20, 30:")
    dll.append(10)
    dll.append(20)
    dll.append(30)
    dll.display_forward()  # Output: 10 <-> 20 <-> 30

    print("\nPrepending 5:")
    dll.prepend(5)
    dll.display_forward()  # Output: 5 <-> 10 <-> 20 <-> 30

    print("\nInserting 15 at index 2:")
    dll.insert_at(2, 15)
    dll.display_forward()  # Output: 5 <-> 10 <-> 15 <-> 20 <-> 30

    print("\nDisplaying backward:")
    dll.display_backward() # Output: 30 <-> 20 <-> 15 <-> 10 <-> 5

    print("\nDeleting 20:")
    dll.delete(20)
    dll.display_forward()  # Output: 5 <-> 10 <-> 15 <-> 30

    print(f"\nFinal List Length: {len(dll)}") # Output: 4
```
