```python
class Node:
  def __init__(self, value):
    self.data = value
    self.next = None

class LinkedList:
  def __init__(self):
    self.head = None
    self.n = 0

  def __len__(self):
    return self.n

#insert
  def insert_head(self, value):
    new_node = Node(value)
    new_node.next = self.head
    self.head = new_node
    self.n = self.n + 1

  def insert_end(self,value):
    new_node = Node(value)

    if self.head == None:
      self.head = new_node
      self.n = self.n + 1
      return

    current = self.head
    while current.next != None:
      current = current.next

    current.next = new_node
    self.n = self.n + 1

  def insert_middle(self, value, position):
    new_node = Node(value)

    if position == 0:
      new_node.next = self.head
      self.head = new_node
      self.n = self.n + 1
      return

    current = self.head
    current_pos = 0

    while current and current_pos < position - 1:
      current = current.next
      current_pos += 1

    if current == None:
      print(f'Position {position} out of bounds')

    new_node.next = current.next
    current.next = new_node
    self.n = self.n + 1

#traverse
  def __str__(self):
    if self.head == None:
      return "List is empty"
    current = self.head

    elements = []

    while current != None:
      elements.append(str(current.data))
      current = current.next

    return '->'.join(elements)


#delete
  def delete_head(self):
    if self.head == None:
      return

    self.head = self.head.next
    self.n = self.n - 1

  def delete_end(self):
    if self.head == None:
      return

    if self.head.next == None:
      self.head = None
      self.n = self.n - 1
      return

    current = self.head
    while current.next.next != None:
      current = current.next

    current.next = None
    self.n = self.n - 1

  def delete_middle(self, position):
    if self.head == None:
      return

    if position == 0:
      self.head = self.head.next
      self.n = self.n - 1
      return

    current = self.head
    current_pos = 0

    while current and current_pos < position - 1:
      current = current.next
      current_pos += 1

    if current == None or current.next == None:
            print(f"Position {position} out of bounds.")
            return

    current.next = current.next.next
    self.n = self.n -1

  #search

  def search(self, value):
    pos = 0

    current = self.head
    while current != None:
      pos = pos +1
      if current.data == value:
        return pos
      current = current.next

    return False

  #update
  def update(self, value, position):
    if self.head == None:
      return "Empty List"
    elif position > self.n:
      return "Position out bounds"

    current = self.head
    current_pos = 0

    while current != None:
      if current_pos == position:
        current.data = value
        return

      current = current.next
      current_pos += 1

  #reverse
  def reverse(self):
    if self.head == None:
      return "Empty List"

    prev = None
    current = self.head

    while current != None:
        next_node = current.next  # Save next node
        current.next = prev       # Flip the pointer
        prev = current            # Shift prev forward
        current = next_node       # Shift current forward

    self.head = prev
```

```python
#Linked List creation and insertion functions

n1 = LinkedList()
n1.insert_head(1)
n1.insert_head(2)
n1.insert_head(3)

n1.insert_middle(4,1)
n1.insert_end(5)

print(n1)
print(len(n1))
```

```python
#Deletion functions

n1.delete_head()
print(n1)
n1.delete_end()
print(n1)
n1.delete_middle(2)
print(n1)
```

```python
#Searching in Linked List
n1.search(3)
n1.search(1)
```

```python
# Updating values in a linked list
n1.update(6, 3)
print(n1)
```

```python
#reversing the linked list
n1.reverse()
print(n1)
```
