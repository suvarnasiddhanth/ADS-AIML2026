Linear search

```python
array = [1,5,8,0,4,7]

def linear_search(key):
  found = False
  for i in range(len(array)):
        if array[i] == key:
            print("Element found at position: ", i)
            found = True

  if found == False:
        print("Element not found")

linear_search(4)
linear_search(9)
```

Binary Search

```python
array = [0,4,6,7,8,9]

def binary_search(key):
    high = len(array)-1
    low = 0
    found = False

    while low<=high:
        mid = (low+high)//2
        if array[mid] == key:
            print("Element found at position: ",mid)
            found = True
            break
        elif array[mid]>key:
            high = mid - 1
        else:
            low = mid + 1

    if not found:
        print("Element not found")

binary_search(8)
binary_search(5)
```
