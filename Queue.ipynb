### Queue implementation using python list -> amortised array

```python
class Queue:
  def __init__(self):
    self.items = []
    self.size = 0

  def isempty(self):
    if self.size == 0:
      return True
    else:
      return self.size

  def enqueue(self, item):
    self.items.append(item)
    self.size += 1

  def dequeue(self):
    if self.size <= 0:
      raise Exception("Queue is empty")
    else:
      self.size -= 1
      return self.items.pop(0)        #O(n) because each deletion leads to shifting of elements to one position left -> Hence start of the array is never empty

  def __repr__(self):
    result = ""
    for item in self.items:
      result += str(item) + "->"
    return result[:-2]

  def test_list(self):
    print(self.items)
```

```python
q = Queue()     #Creating a queue

#Enqueue
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.enqueue(40)
```

```python
#display Queue
print(q)
```

```python
#Dequeue

q.dequeue()
print(q)
```

```python
#Queue length check
q.isempty()
```

```python
q.test_list()
```

### Static array implementation of Queue


```python
class staticQueue:
  def __init__(self, capacity):
    self.front = 0
    self.rear = -1
    self.size = 0
    self.capacity = capacity
    self.items = [None] * capacity

  def isempty(self):
    return self.size == 0

  def enqueue(self, item):
    if self.rear == self.capacity - 1:
      print("Stack Overflow")

    self.rear += 1
    self.items[self.rear] = item
    self.size += 1

  def dequeue(self):
    if self.rear == -1:
      print("Stack Underflow")

    removed_item = self.items[self.front]
    self.items[self.front] = None
    self.front += 1

    print(removed_item)

  def __repr__(self):
    result = ""

    for items in self.items:
      result += str(items) + "->"
    return result[:-2]
```

```python
queue = staticQueue(4)

queue.enqueue(10)
queue.enqueue(20)
queue.enqueue(30)
queue.enqueue(40)
queue.enqueue(50)
```

```python
print(queue)
```

```python
queue.dequeue()
```

```python
print(queue)    #Array is still full since front index is updated by 1 but the size still remains the same -> working is different from pop()
```

### Implementation of Queues using linked list

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

class ListQueue:
  def __init__(self):
    self.front = None
    self.rear = None
    self.size = 0

  def isempty(self):
    if self.size == 0:
      raise IndexError("Queue Underflow")
    else:
      print(self.size)

  def enqueue(self, data):          #insertion always from tail
    newNode = Node(data)

    if self.size == 0:
      self.front = newNode
      self.rear = newNode

    else:
      self.rear.next = newNode
      self.rear = newNode

    self.size += 1

  def dequeue(self):
    if self.front == None:
      self.rear = None
      raise IndexError("Queue Underflow")
    else:
      removed_data = self.front.data
      self.front = self.front.next

    self.size -= 1
    return removed_data

  def __repr__(self):
    if self.size == 0:
      raise IndexError("Queue Underflow")

    result = []
    current = self.front

    while current is not None:
      result.append(str(current.data))
      current = current.next

    return "->".join(result)
```

```python
L_Queue = ListQueue()

L_Queue.enqueue(10)
L_Queue.enqueue(20)
L_Queue.enqueue(30)
L_Queue.enqueue(40)
L_Queue.enqueue(50)
```

### Circular Queue

```python
class CircularQueue:
  def __init__(self, capacity):
    self.front = 0
    self.rear = -1
    self.capacity = capacity
    self.items = [None] * capacity
    self.size = 0

  def isempty(self):
    return self.size == 0
    if self.size == 0:
      raise IndexError("Queue is empty - Underflow")
    else:
      print(self.size)

  def isfull(self):
    return self.size == self.capacity


  def enqueue(self, item):
    if self.isfull():
      raise IndexError("Queue is full - Overflow")

    self.rear = (self.rear + 1) % self.capacity
    self.items[self.rear] = item
    print(self.front, self.rear)
    self.size += 1

  def dequeue(self):
    if self.isempty():
      raise IndexError("Queue is empty - Underflow")

    deleted_item = self.items[self.front]
    self.items[self.front] = None

    if self.front == self.rear:
      self.front = -1
      self.rear = -1
    else:
      self.front = (self.front + 1) % self.capacity

    self.size -= 1
    return deleted_item

  def __repr__(self):
    if self.isempty():
      raise IndexError("Queue is empty - Underflow")

    result = []

    current = self.front
    count = 0

    while count < self.size:
      result.append(str(self.items[current]))
      current = (current + 1) % self.capacity
      count += 1

    return "->".join(result)
```

```python
C_queue = CircularQueue(5)

C_queue.enqueue(10)
C_queue.enqueue(20)
C_queue.enqueue(30)
C_queue.enqueue(40)
C_queue.enqueue(50)

print(C_queue)
```

```python
C_queue.dequeue()
print(C_queue)
```
