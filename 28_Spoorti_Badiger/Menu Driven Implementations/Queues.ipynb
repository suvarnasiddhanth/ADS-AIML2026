```python
class Queue:
  def __init__(self):
    self.que = []

  def enqueue(self,element):
    self.que.append(element)

  def dequeue(self):
    if not self.que:
      print("Queue is empty")
      return None
    deletedElement = self.que.pop(0)

    return deletedElement

  def display(self):
    for i in range(len(self.que)):
      print(f"{i} -> {self.que[i]}")


def main():
  queue = Queue()
  #menu driven implementaion
  while True:
    print("1. Enqueue")
    print("2. Dequeue")
    print("3. Display")
    print("4. Exit")
    choice = int(input("Enter your choice: "))
    if choice == 1:
      element = int(input("Enter the element to enqueue: "))
      queue.enqueue(element)
    elif choice == 2:
      deletedElement = queue.dequeue()
      if deletedElement is not None:
        print(f"Deleted element: {deletedElement}")
    elif choice == 3:
      queue.display()
    elif choice == 4:
      print("Existing...")
      break
    else:
      print("Invalid choice")


if __name__ =="__main__":
  main()
```
