```python
class BinarySearch:
  def __init__(self):
    self.arr = []

  def Binsearch(self, arr, target):
    self.arr = arr
    low = 0
    high = len(self.arr) - 1
    while low <= high:
      mid = (low + high) // 2
      if target == self.arr[mid]:
        return mid
      elif target < self.arr[mid]:
        high = mid - 1
      else:
        low = mid + 1
    return None


def main():
  arr = [22, 56, 89, 100, 124]
  Search = BinarySearch()
  found = Search.Binsearch(arr, 100)
  if found is not None:
    print("The required target is present at the index: ", found)
  else:
    print("The required element is not present in the given array")

if __name__ == "__main__":
  main()
```
