```python
class InsertionSort:
  def __init__(self,arr):
    self.arr = arr

  def insertion_sort(self):
    for i in range(len(self.arr)):
      key = self.arr[i]
      j=i-1
      while j>=0 and arr[j] > key:
        self.arr[j+1] = self.arr[j]
        j=j-1
      self.arr[j+1] = key
    return self.arr


arr = [4,9,10,0,6,8]
sorted = InsertionSort(arr)
sorted.insertion_sort()
print("The sorted array is: ",sorted.arr)
```
