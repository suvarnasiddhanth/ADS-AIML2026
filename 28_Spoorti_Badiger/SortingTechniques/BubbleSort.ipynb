```python
class BubbleSort:
  def __init__(self,arr):
    self.arr = arr

  def sort(self):
    n = len(self.arr)
    for i in range(n):
      for j in range(0,n-i-1):
        if self.arr[j+1] < self.arr[j]:
          self.arr[j],self.arr[j+1] = self.arr[j+1],self.arr[j]

    return self.arr



array = [2,6,9,4,0,8]
sorted = BubbleSort(array)
print("The sorted aray is:",sorted.sort())
```
