```python
class Queue:
  def __init__(self):
    self.que = []

  def enqueue(self,element):
    self.que.append(element)

  def dequeue(self):
    if not self.que:
      print("Queue is empty")
      return None
    deletedElement = self.que.pop(0)

    return deletedElement

  def display(self):
    for i in range(len(self.que)):
      print(f"{i} -> {self.que[i]}")


def main():
  queue = Queue()
  queue.enqueue(3)
  queue.enqueue(4)
  queue.enqueue(9)
  queue.enqueue(10)
  print("The initial queue: ")
  queue.display()
  print("Deleted element is:",queue.dequeue())
  queue.display()


if __name__ == "__main__":
  main()
```
