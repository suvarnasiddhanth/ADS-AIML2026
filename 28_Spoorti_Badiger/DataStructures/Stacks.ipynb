```python
class stacks:
  def __init__(self):
    self.stack = []

  def push(self,element):
    self.stack.append(element)

  def pop(self):
    if not self.stack:
      print("Underflow")
      return

    deletedElement = self.stack.pop()
    return deletedElement

  def peek(self):
    if not self.stack:
      print("Stack is empty")
      return

    return self.stack[-1]

  def isempty(self):
    if not self.stack:
      return True
    else:
      return False

  def size(self):
    return len(self.stack)

  def display(self):
    if not self.stack:
      print("Stack is empty")

    print(self.stack)



def main():
  stack = stacks()
  stack.display()
  stack.push(10)
  stack.push(30)
  stack.push(66)
  stack.push(99)
  stack.push(100)
  stack.display()
  stack.pop()
  stack.pop()
  stack.display()

if __name__ == "__main__":
  main()
```
