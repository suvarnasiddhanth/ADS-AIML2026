# Singly Linked List

```python
class Node:
  def __init__(self, value):
    self.data = value
    self.next = None

class LinkedList:
  def __init__(self):
    self.head = None
    self.n = 0

  def __len__(self):
    return self.n

#insert
  def insert_head(self, value):
    new_node = Node(value)
    new_node.next = self.head
    self.head = new_node
    self.n = self.n + 1

  def insert_end(self,value):
    new_node = Node(value)

    if self.head == None:
      self.head = new_node
      self.n = self.n + 1
      return

    current = self.head
    while current.next != None:
      current = current.next

    current.next = new_node
    self.n = self.n + 1

  def insert_middle(self, value, position):
    new_node = Node(value)

    if position == 0:
      new_node.next = self.head
      self.head = new_node
      self.n = self.n + 1
      return

    current = self.head
    current_pos = 0

    while current and current_pos < position - 1:
      current = current.next
      current_pos += 1

    if current == None:
      print(f'Position {position} out of bounds')

    new_node.next = current.next
    current.next = new_node
    self.n = self.n + 1

#traverse
  def __str__(self):
    if self.head == None:
      return "List is empty"
    current = self.head

    elements = []

    while current != None:
      elements.append(str(current.data))
      current = current.next

    return '->'.join(elements)


#delete
  def delete_head(self):
    if self.head == None:
      return

    self.head = self.head.next
    self.n = self.n - 1

  def delete_end(self):
    if self.head == None:
      return

    if self.head.next == None:
      self.head = None
      self.n = self.n - 1
      return

    current = self.head
    while current.next.next != None:
      current = current.next

    current.next = None
    self.n = self.n - 1

  def delete_middle(self, position):
    if self.head == None:
      return

    if position == 0:
      self.head = self.head.next
      self.n = self.n - 1
      return

    current = self.head
    current_pos = 0

    while current and current_pos < position - 1:
      current = current.next
      current_pos += 1

    if current == None or current.next == None:
            print(f"Position {position} out of bounds.")
            return

    current.next = current.next.next
    self.n = self.n -1

  #search

  def search(self, value):
    pos = 0

    current = self.head
    while current != None:
      pos = pos +1
      if current.data == value:
        return pos
      current = current.next

    return False

  #update
  def update(self, value, position):
    if self.head == None:
      return "Empty List"
    elif position > self.n:
      return "Position out bounds"

    current = self.head
    current_pos = 0

    while current != None:
      if current_pos == position:
        current.data = value
        return

      current = current.next
      current_pos += 1

  #reverse
  def reverse(self):
    if self.head == None:
      return "Empty List"

    prev = None
    current = self.head

    while current != None:
        next_node = current.next  # Save next node
        current.next = prev       # Flip the pointer
        prev = current            # Shift prev forward
        current = next_node       # Shift current forward

    self.head = prev
n1 = LinkedList()
n1.insert_head(1)
n1.insert_head(2)
n1.insert_head(3)

n1.insert_middle(4,1)
n1.insert_end(5)

print(n1)
print(len(n1))

n1.delete_head()
print(n1)
n1.delete_end()
print(n1)
n1.delete_middle(2)
print(n1)
n1.search(3)
n1.search(1)
n1.update(6, 3)
print(n1)
n1.reverse()
print(n1)
```

# Doubly Linked List

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.prev = None
    self.next = None
class DoublyLinkedList:
    """A class representing a doubly linked list."""
    def __init__(self):
        self.head = None
        self.tail = None
        self.length = 0

    def __len__(self):
        """Return the total number of nodes in O(1) time."""
        return self.length

    def prepend(self, data):
        """Insert a new node at the beginning of the list."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            new_node.next = self.head
            self.head.prev = new_node
            self.head = new_node

        self.length += 1

    def append(self, data):
        """Insert a new node at the end of the list."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            new_node.prev = self.tail
            self.tail = new_node

        self.length += 1

    def insert_at(self, index, data):
        """Insert a value at a specific position."""
        if index < 0 or index > self.length:
            raise IndexError("Index out of bounds")

        if index == 0:
            self.prepend(data)
            return
        if index == self.length:
            self.append(data)
            return

        new_node = Node(data)
        current = self.head

        # Traverse to the node currently occupying the target index
        for _ in range(index):
            current = current.next

        # Link the new node into the list
        new_node.prev = current.prev
        new_node.next = current
        current.prev.next = new_node
        current.prev = new_node

        self.length += 1

    def delete(self, value):
        """Remove the first occurrence of a value from the list."""
        current = self.head

        while current is None:
            if current.data == value:
                break
            current = current.next

        if current is None:
            return False  # Value not found

        # Update Head pointer if removing the first node
        if current == self.head:
            self.head = current.next
        else:
            current.prev.next = current.next

        # Update Tail pointer if removing the last node
        if current == self.tail:
            self.tail = current.prev
        else:
            current.next.prev = current.prev

        self.length -= 1
        return True

    def display_forward(self):
        """Print all elements from head to tail."""
        elements = []
        current = self.head
        while current:
            elements.append(str(current.data))
            current = current.next
        print(" <-> ".join(elements) if elements else "Empty List")

    def display_backward(self):
        """Print all elements from tail to head."""
        elements = []
        current = self.tail
        while current:
            elements.append(str(current.data))
            current = current.prev
        print(" <-> ".join(elements) if elements else "Empty List")


# --- Demonstration ---
if __name__ == "__main__":
    dll = DoublyLinkedList()

    print("Appending 10, 20, 30:")
    dll.append(10)
    dll.append(20)
    dll.append(30)
    dll.display_forward()  # Output: 10 <-> 20 <-> 30

    print("\nPrepending 5:")
    dll.prepend(5)
    dll.display_forward()  # Output: 5 <-> 10 <-> 20 <-> 30

    print("\nInserting 15 at index 2:")
    dll.insert_at(2, 15)
    dll.display_forward()  # Output: 5 <-> 10 <-> 15 <-> 20 <-> 30

    print("\nDisplaying backward:")
    dll.display_backward() # Output: 30 <-> 20 <-> 15 <-> 10 <-> 5

    print("\nDeleting 20:")
    dll.delete(20)
    dll.display_forward()  # Output: 5 <-> 10 <-> 15 <-> 30

    print(f"\nFinal List Length: {len(dll)}") # Output: 4
```

# Food truck problem

```python
class CustomerNode:
    """Represents an individual customer in the queue."""
    def __init__(self, name):
        self.name = name
        self.next = None  # Pointer to the next customer

class FoodTruckQueue:
    """Manages customer distribution between two food trucks."""
    def __init__(self):
        self.head = None
        self.tail = None

    def add_customer(self, name):
        """Adds a new customer to the end of the singly linked list."""
        new_customer = CustomerNode(name)
        if not self.head:
            self.head = new_customer
            self.tail = new_customer
        else:
            self.tail.next = new_customer
            self.tail = new_customer

    def distribute_and_serve(self):
        """Distributes customers between Truck A and Truck B without bias."""
        if not self.head:
            print("No customers in line!")
            return

        truck_a_queue = []
        truck_b_queue = []

        current = self.head
        serve_to_truck_a = True  # Flag to alternate between trucks

        # Traverse the linked list to distribute customers
        while current:
            if serve_to_truck_a:
                truck_a_queue.append(current.name)
            else:
                truck_b_queue.append(current.name)

            # Alternate the flag for the next customer
            serve_to_truck_a = not serve_to_truck_a
            current = current.next

        # Display the unbiased serving order
        print("TRUCK A QUEUE:", " -> ".join(truck_a_queue))
        print("TRUCK B QUEUE:", " -> ".join(truck_b_queue))


# --- Simulation with 10 Peak Hour Customers ---
if __name__ == "__main__":
    line = FoodTruckQueue()

    # Simulating 10 customers arriving at peak hour
    for i in range(1, 11):
        line.add_customer(f"Customer {i}")

    print("--- Distributing 10 Peak Hour Customers Alternately ---")
    line.distribute_and_serve()
```
