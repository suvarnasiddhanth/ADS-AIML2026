```python
class Queue:
  def enque(self,qu,data):
    qu.append(data)

  def dequeue(self,qu):
    if len(qu)==0:
      print("Queue is empty")
    rem=qu.pop(0)
    print("The dequed element is:",rem)

  def peek(self,qu):
    return qu[0]

  def size(self,qu):
    if len(qu)==0:
      print("the list is empty")
    print("The size is :",len(qu))


if __name__=="__main__":
  qu=Queue()
  li=[10,20,30,40,50,60,70,80]

  print(li)Untitled2
  qu.dequeue(li)
  print(li)
  print(qu.peek(li))
  qu.enque(li,90)
  print(li)
  qu.size(li)
```
