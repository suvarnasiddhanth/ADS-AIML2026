```python
class Stack:
  def push(self,st,data):
    st.append(data)

  def pop(self,st):
    if len(st)==0:
      print("NOthing to pop")
    else:
      var=st.pop()
      print("The popped element is: ",var)

  def top(self,st):
    if len(st)>0:
      return st[len(st)-1]


  def isEmpty(self,st):
    if len(st)==0:
      print("The stack is empty")
    return 0

  def size(self,st):
    return len(st)


if __name__=="__main__":
  st=Stack()
  s=[10,20,30,40,50,60,70,80,90]

  print(s)
  st.pop(s)
  val=100
  st.push(s,val)
  print("The pushed element is ",s[-1])
  print(st.isEmpty(s))
  print("the topmost element of the stack is :",st.top(s))
  print("New stack",s)
  print("the length of the stack is :",st.size(s))


```
