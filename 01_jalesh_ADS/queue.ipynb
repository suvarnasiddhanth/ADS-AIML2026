```python
# enqueue
queue=[10,20,30]
queue.append(40)
print(queue)
```

```python
# specific postion
queue=[10,20,30]
queue.insert(0,50)
print(queue)
```

```python
#Enqueue
queue=[]
queue.append(10)
queue.append(90)
queue.append(70)
print("queue",queue)
```

```python
# dequeue
queue = [10, 20, 30]

queue.pop(0)

print(queue)
```

```python
# peek
queue=[10,20,30]
print(queue[0])
```

```python
# len
queue = [10, 20, 30]

print(len(queue))
```

```python
# queue is empty
queue = []

if len(queue) == 0:
    print("Queue is empty")
else:
    print("Queue is not empty")
```
