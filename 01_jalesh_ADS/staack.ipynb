```python
class stack:
    def push(self,st,data):
        st.append(data)
    
    def pop(self,st):
        if len(st)>0:
            ans=st.pop()
            print("element is removed",ans)
        else:
            print("stack is empty")
    def top(self,st):
        if len(st)>0:
            return st[-1]
    def isEmpty(self,st):
        if len(st)==0:
            return 1
        else:
            return 0
    def size(self,st):
        return len(st)
#if __name__=="__main__"
st=stack()
s=["jay","rahul","rohit","sachin"]
print(s)
print(st.size(s))
st.pop(s)

print("top=",st.top(s))
print(st.isEmpty(s))

st.push(s,"vimal kant")
print(s)
print(st.top(s))
```
