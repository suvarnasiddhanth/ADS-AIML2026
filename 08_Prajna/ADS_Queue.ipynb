##QUEUE IMPLEMENTATION

```python
class Queue:
  def __init__(self):
    self.queue = []

  def enqueue(self, element):
    self.queue.append(element)

  def dequeue(self):
    if self.isEmpty():
      return "Queue is empty"
    return self.queue.pop(0)

  def peek(self):
    if self.isEmpty():
      return "Queue is empty"
    return self.queue[0]

  def isEmpty(self):
    return len(self.queue) == 0

  def size(self):
    return len(self.queue)

# Create a queue
myQueue = Queue()

myQueue.enqueue('P')
myQueue.enqueue('Q')
myQueue.enqueue('R')

print("Queue: ", myQueue.queue)
print("Peek: ", myQueue.peek())
print("Dequeue: ", myQueue.dequeue())
print("Queue after Dequeue: ", myQueue.queue)
print("isEmpty: ", myQueue.isEmpty())
print("Size: ", myQueue.size())
```

## CIRCULAR QUEUE IMPLEMENTATION

```python
class CircularQueue:
    def __init__(self,size):
        self.size = size
        self.queueArray = [None] * size
        self.front = -1
        self.rear = -1

    def isEmpty(self):
        return self.front == -1 and self.rear == -1

    def enqueue(self,data):
        if (self.rear + 1) % self.size == self.front:
            print("Queue is full")
            return
        if self.front == -1:
            self.front = 0

        self.rear = (self.rear + 1) % self.size
        self.queueArray[self.rear] = data

    def dequeue(self):
        if self.front==-1:
            print("Queue is empty")
            return None

        data = self.queueArray[self.front]
        self.queueArray[self.front] = None

        if self.front == self.rear:
            self.front = -1
            self.rear = -1
        else:
            self.front = (self.front+1)%self.size

        return data

    def peek(self):
        if not self.isEmpty():
            return self.queueArray[self.front]
        else:
            print("Queue is empty. No peek value.")
            return -1  # Assuming -1 represents an empty value

    def display(self):
        print("Queue:",self.queueArray)

myQueue = CircularQueue(4)

myQueue.enqueue(2)
myQueue.enqueue(4)
myQueue.enqueue(8)
myQueue.enqueue(14)

myQueue.display()

print("Peek:", myQueue.peek())

print("Dequeued:", myQueue.dequeue())
print("Dequeued:", myQueue.dequeue())

myQueue.display()

myQueue.enqueue(24)
myQueue.enqueue(44)

myQueue.display()
```
