## SINGLY LINKED LIST IMPLEMENTATION

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None

head=None
n = 0

def insertFront(data):
    global head, n
    newNode=Node(data)
    if head is None:
        head=newNode
    else:
        newNode.next=head
        head=newNode
    n = n+1

def traverse():
    curr = head

    while curr is not None:
        print(curr.data, end="")

        if curr.next is not None:
            print(" → ", end="")

        curr = curr.next

    print()

def insertEnd(data):
    global head, n
    newNode=Node(data)
    if head is None:
        head=newNode
    else:
        curr=head
        while curr.next is not None:
            curr=curr.next
        curr.next=newNode
    n = n+1

def insertAtPosition(data, position):
    global head, n
    newNode=Node(data)
    if position==0:
        newNode.next=head
        head=newNode
    else:
        curr=head
        for i in range(position-1):
            curr=curr.next

        newNode.next=curr.next
        curr.next=newNode
    n = n+1

def deleteAtFront():
    global head, n
    if head is None:
        return
    else:
        head=head.next
    n = n-1

def deleteAtEnd():
    global head, n
    if head is None:
        return
    else:
        curr=head
        while curr.next.next !=None:
            curr=curr.next
        curr.next=None
    n = n-1

def deleteAtPos(pos):
    global head, n
    if pos==0:
        head=head.next
    else:
        curr=head
        for i in range(pos-1):
            curr=curr.next
        curr.next=curr.next.next
    n = n-1

def search(data):
    curr=head
    pos=0
    while curr is not None:
        if curr.data==data:
            print(data,pos)
            return
        else:
            pos+=1
            curr=curr.next
    return False

def reverse():
    global head
    prev,curr=None,head
    while curr is not None:
        next=curr.next
        curr.next=prev
        prev=curr
        curr=next
    head=prev

def length():
    return n


insertFront(2)
insertFront(4)
insertEnd(6)
insertEnd(8)
insertFront(10)
insertAtPosition(14,4)
#deleteAtFront()
#deleteAtEnd()
#deleteAtPos(2)
insertEnd(18)
#search(8)
traverse()
print("Reverse: ")
reverse()
traverse()
print("Length:", length())
```

## DOUBLY LINKED LIST IMPLEMENTATION



```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        self.prev=None

head=None
n=0

def insertFront(data):
    global head, n
    newNode = Node(data)
    if head is None:
        head=newNode
    else:
        newNode.next=head
        head.prev=newNode
        head=newNode
    n = n+1

def traverse():
    if head is None:
        print("List is empty")
        return

    curr = head

    while curr is not None:
        print(curr.data, end="")

        if curr.next is not None:
            print(" → ", end="")

        curr = curr.next

    print()

def insertEnd(data):
    global head, n
    newNode=Node(data)
    if head is None:
        head=newNode
        n = n+1
        return
    curr=head
    while curr.next is not None:
        curr=curr.next
    curr.next=newNode
    newNode.prev=curr
    n=n+1

def deleteFront():
    global head, n
    if head is None:
        return
    elif head.next is None:
        head=None
    else:
        curr=head.next
        curr.prev=None
        head.next=None
        head=curr
    n=n-1

def deleteEnd():
    global head, n
    if head is None:
        return
    elif head.next is None:
        head=None
    else:
        curr=head
        while curr.next.next is not None:
            curr=curr.next
        curr.next=None
    n=n-1

def insertAtPos(data,pos):
    global head,n
    newNode=Node(data)
    if pos==0:
        newNode.next=head
        if head is not None:
            head.prev=newNode
        head=newNode
        n=n+1
        return

    curr=head
    for i in range(pos-1):
      curr=curr.next
      #insert at end
      if curr is None:
        print("Invalid Index , Node is Inserted at End")
        insertEnd(data)
        return
    if curr.next is None:
      insertEnd(data)
      return
    #insert in middle
    nextNode=curr.next
    curr.next=newNode
    newNode.prev=curr
    newNode.next=nextNode
    nextNode.prev=newNode
    n=n+1

def deleteAtPos(pos):
    global head, n
    if pos==0:
        deleteFront()
        return

    curr=head
    for i in range(pos-1):
      curr=curr.next
      if curr is None:
        print("Invalid Index")
        return
    #if deleting last node
    if curr.next is None:
      deleteEnd()
      return
    nextNode=curr.next.next
    #delete middle node
    curr.next=nextNode
    nextNode.prev=curr
    n=n-1

def length():
    return n

insertFront(2)
insertFront(4)
insertFront(6)
insertEnd(8)
insertEnd(10)
deleteFront()
deleteEnd()
insertAtPos(44,1)
insertAtPos(48,2)
insertAtPos(54,100)
deleteAtPos(2)
deleteAtPos(3)
traverse()

print("Length:", length())
```
