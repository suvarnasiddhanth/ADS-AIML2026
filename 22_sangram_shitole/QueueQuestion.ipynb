###Depending upon the length, increase the queue size. When you remove an element, reduce the size. Dynamically increase or decrease the size depending on the requirement. A maximum capacity should increase or decrease.

```python
class DynamicQueue:
  def __init__(self,capacity=4):
    self.capacity=capacity
    self.queue=[None]*capacity
    self.head=0
    self.rear=0
    self.front=0
    self.tail=0
    self.size=0

  def enqueue(self,x):
      #Queue is full
      if self.size ==self.capacity:
        self.resize(self.capacity*2)


      self.queue[self.rear]=x
      self.rear=(self.rear+1)%self.capacity
      self.size+=1

  def dequeue(self):
    #Queue is empty
    if self.size==0:
      print("Queue is empty")
      return

    x=self.queue[self.front]
    self.queue[self.front]=None
    self.front=(self.front +1)%self.capacity
    self.size=-1

    #reduce capacity if queue is mostly empty
    if self.size >0 and self.size <=self.capacity//4:
      self.resize(self.capacity//2)

    return x

  def resize(self,new_capacity):
    new_queue=[None]* new_capacity
    #copy element in correct queue order
    for i in range(self.size):
      new_queue[i]=self.queue[(self.front+1)%self.capacity]


    self.queue=new_queue
    self.front=0
    self.rear=self.size
    self.capacity=new_capacity

  def display(self):
    print("Queue:", end=' ')
    for i in range(self.size):
      print(self.queue[self.front +1] % self.capacity,end =" ")

      print()
      print("self :",self.size)
      print("capacity :" , self.capacity)

q = DynamicQueue(4)

q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.enqueue(40)

q.display()

q.enqueue(50)

q.display()

q.dequeue()
q.dequeue()
q.dequeue()
q.dequeue()

q.display()


```
